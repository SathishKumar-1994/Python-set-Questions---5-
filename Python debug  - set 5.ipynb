```python
#8.What is the issue with this function?

def sum_numbers(numbers):
    total = 1
    for num in numbers:
        total += num
    return total
```

```python
def sum_numbers(numbers):
    total = 1
    for num in str(numbers):
        total += int(num)
    return total
    
print(sum_numbers(156))
```
